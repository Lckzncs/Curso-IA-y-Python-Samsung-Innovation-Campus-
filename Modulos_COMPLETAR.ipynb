# Módulos en Python

Hasta ahora siempre hemos utilizado _jupyter notebooks_ para escribir nuestro código Python.

En realidad, los programas Python normalmente se almacenan en ficheros de texto con extensión `.py`. A estos ficheros se les llama _scripts_ Python y en ellos solo puede aparecer código Python (con comentarios). En un _script_ no hay celdas de código o documentación markdown.

Para trabajar con un _script_ Python no podemos usar jupyter; debemos usar al menos:

- un editor simple (preferentemente que coloree los programas Python) para **editar** los _scripts_
- el REPL de Python (`python` o `ipython`) para **ejecutar** los _scripts_

Alternativamente, podemos usar herramientas que permitan tanto ejecutar como editar los _scripts_:

- un editor con soporte para Python, como Visual Studio Code (VS Code)
- un IDE (Integrated Development Environment), como PyCharm

## _Scripts_ Python

Para terminar, vamos a escribir un _script_ más grande que define una función `resuelve_ecuacion` para resolver una ecuación de segundo grado. Copia el siguiente código en un fichero `raices.py`:

```python
# script (programa) para calcular las raíces de una ecuación de segundo grado

import math

def resuelve_ecuacion(a,b,c):
    """
    Resuelve una ecuacion de segundo grado a x^2 + bx + c = 0.

    Parameters
    ----------
    a : float
        Coeficiente de segundo grado. No puede ser cero.
    b : float
        Coeficiente de primer grado.
    c : float
        Término independiente.

    Returns
    -------
    Una tupla con las raíces reales.

    Throws
    ------
    ValueError si no hay dos soluciones reales.
    """

    if a == 0:
        print("error: la ecuación no es de segundo grado")
    else:
        discriminante = b ** 2 - 4 * a * c
        if discriminante < 0:
            print("error: las raíces son complejas")
        else:
            raiz_discriminante = math.sqrt(discriminante)
            raices = ((-b + raiz_discriminante) / (2 * a),
                      (-b - raiz_discriminante) / (2 * a))
            return raices


a = float(input("dame el coeficiente a: "))

```

Como ves, un _script_ puede definir funciones y usarlas. También puede usar funciones de las bibliotecas de Python (`import math`). Además, el código Python de un _script_ puede (¡y debe!) estar documentado. En esta ocasión hemos utilizado una _docstring_ para documentar la función `resuelve_ecuacion`.

## Módulos Python

Un módulo Python es similar a un _script_ Python, pero no está pensado para ser ejecutado, sino importado. Ya hemos usado varios módulos Python (`math`, `random`, `json`).

Un módulo Python contiene un conjunto de definiciones quer podemos usar desde otros módulos, _scripts_ o _notebooks_.

Podemos convertir el anterior _script_ Python en un módulo simplemente eliminado el código que se ejecuta y dejando solo las definiciones (en nuestro caso, la función `resuelve_ecuacion`).

Borra el contenido anterior del fichero `raices.py` y copia el siguiente código:

```python
# módulo para calcular las raíces de una ecuación de segundo grado

import math


def resuelve_ecuacion(a, b, c):
    """
    Resuelve una ecuacion de segundo grado a x^2 + bx + c = 0.

    Parameters
    ----------
    a : float
        Coeficiente de segundo grado. No puede ser cero.
    b : float
        Coeficiente de primer grado.
    c : float
        Término independiente.

    Returns
    -------
    Una tupla con las raíces reales.

    Throws
    ------
    ValueError si no hay dos soluciones reales.
    """

    if a == 0:
        raise ValueError("la ecuación no es de segundo grado")
    else:
        discriminante = b ** 2 - 4 * a * c
        if discriminante < 0:
            raise ValueError("las raíces son complejas")
        else:
            raiz_discriminante = math.sqrt(discriminante)
            raices = ((-b + raiz_discriminante) / (2 * a),
                      (-b - raiz_discriminante) / (2 * a))
            return raices
```

Observa que `raices.py` ahora no ejecuta ninguna acción, no pide coeficientes ni resuelve una ecuación, simplemente define una función. Si ejecutas `raices.py` desde VS Code (o desde el REPL), verás que no ocurre nada.

## Módulo y _script_ Python

Al escribir un fichero Python `.py`, no estamos obligados a escoger su papel entre _script_ y módulo. Podemos escribir el fichero `.py` y dejar que sea el usuario quien decida si lo quiere usar como un _script_ (ejecutar las acciones que contiene) o como un módulo (importar las definiciones que contiene).

Añade al final de `raices.py` el siguiente código:

```python
if __name__ == "__main__":
    # esto se ejecuta al usarlo como script, pero no al importarlo
    a = float(input("dame el coeficiente a: "))
    b = float(input("dame el coeficiente b: "))
    c = float(input("dame el coeficiente c: "))
    r1, r2 = resuelve_ecuacion(a, b, c)
    print("las raíces son" ,r1, r2)
```

Observa que lo que estamos haciendo es condicionar la ejecución de las acciones del _script_, que se encargan de preguntar al usuario por los coeficientes de la ecuación y la resuelven.

- Si ejecutamos `raices.py` desde VS Code o desde el REPL, se ejecutarán las acciones del _script_
- Si importamos `raices`, sus definiciones pasan a estar disponibles, pero no se ejecuta el _script.

A veces el nombre de un módulo es demasiado largo y resulta molesto escribir su nombre delante de todas sus definiciones. Para evitar esto, podemos utilizar la siguiente forma de `import`:

```python
import nombre_modulo_completo as nombre_modulo_abreviado
```

donde `nombre_modulo_abreviado` es un nombre que nos inventamos nosotros. Para algunos módulos, el nombre abreviado está estandarizado.

````python
import numpy as np
import pandas as pd
import matplotlib as plt
````

A veces un módulo exporta demasiadas definiciones y solo nos interesa utilizar unas pocas. En tal caso, podemos usar una forma de `import` que menciona las definiciones que queremos usar:

```python
from nombre_modulo import nombre_1, nombre_2, ...
```

Una ventaja de esta forma de `import` es que las definiciones importadas (`nombre_1`, `nombre_2`, ...) se utilizan sin precederlas del nombre del módulo.

Por ejemplo, del módulo `math` podemos importar `pi`, `cos`, `sin` y `sqrt`:

````python
from math import pi, cos, sin , sqrt
````

````python
pi
````

````python
sqrt(pi)
````

Finalmente, es posible importar todas las definiciones de un módulo sin mencionarlas:

```python
from nombre_modulo import *
```

Esto permite usar todas las definiciones sin precederlas del nombre del módulo. Sin embargo, esta forma de importación se considera mal estilo y está **desaconsejada**.

## EJERCICIO 1

Queremos implementar un programa que simule jugar a los dados contra la máquina. Al inicio del juego se escogerá el número de caras del dado que se utilizará (4,6,8,10,12 o 20) y también cuántos dados se van a lanzar en total.

Se le pedirá al jugador humano una confirmación para empezar el juego, mostrando la cadena "¿Listo para comenzar el juego? Presione alguna tecla para continuar"

Una vez hecha la tirada, cada jugador podrá decidir si conservar el valor del dado o volver a lanzarlo una sola vez más. Se utilizará '-' para indicar que se deja el dado como está y 'r para indicar que se relanza el dado.
La máquina sigue dos estrategias disintas: la primera estrategia, con una probabilidad del 25%, consiste en no cambiar ningún dado; la segunda estrategia, con una probabilidad del 75%, consiste en relanzar todo dado con valor menor o igual a la mitad del valor máximo del tipo de dado. 
Al jugador humano se le mostrará el resultado de su primera tirada para que pueda decidir qué hacer. La máquina decidirá sin que el jugador humano vea su resultado hasta el final.

Finalmente, se sumará el valor de todos los dados y el jugador con mayor valor será el ganador.

## Indicaciones de diseño

Se debe definir un módulo 'dados.py' con las siguientes funciones:

**tipo_dado_valido(t)**: devuelve True si t es un número de caras válido para los dados.

**lanzar_dados(n,t)**: devuelve una lista con el resultado de lanzar n dados de tipo t. En caso de que no se indique el tipo se lanzarán dados de 6 caras. Si se indica un tipo de dado incorrecto se devolverá un error de tipo *ValueError*.

**relanzar_decision(decision,lista_dados, t)**: devuelve una nueva lista con el contenido de lista_dados tras relanzar o no los dados de t caras indicados en la cadena decision. Em dicha cadena '-' indica no relanzar, y 'r' sí hacerlo.

**encontrar(lista_datos, valor_umbral)**: devuelve una lista con las posiciones de los dados cuyo valor es menor o igual a valor_umbral.

**relanzar_indices(lista_indices, lista_dados, t)**: devuelve una nueva lista con el contenido de  lista_dados tras relanzar los dados de t caras de las posiciones dadas en lista_indices.

**puntuacion(lista_dados)**: devuelve la puntuacion final de la lista de dados dada (suma de todos los valores).

````python
import dados
import random

print("¡Hola! Vamos a jugar a los dados.")

t = int(input("¿De cuántas caras quieres el dado (4, 6, 8, 10, 12 o 20)? "))
n = int(input("¿Cuántos dados tiramos? "))

input("¿Listo para empezar? Pulsa Enter para lanzar los dados...\n")

if not dados.tipo_dado_valido(t):
    print("Ese tipo de dado no es valido. Asegúrate de poner uno de la lista.")
else:
    dados_jugador = dados.lanzar_dados(n, t)
    dados_maquina = dados.lanzar_dados(n, t)

    print(f"\nTe ha salido esto en tu tirada: {dados_jugador}")
    
    print("Pon '-' para quedarte el dado o 'r' para volver a tirarlo.")
    print("Ejemplo: si tienes 3 dados y quieres cambiar solo el del medio, escribe: -r-")
    decision = input("Dime tu jugada: ")
    
    dados_jugador_final = dados.relanzar_decision(decision, dados_jugador, t)

    if random.random() < 0.75:
        umbral = t / 2
        indices_malos = dados.encontrar(dados_maquina, umbral)
        dados_maquina_final = dados.relanzar_indices(indices_malos, dados_maquina, t)
    else:
        dados_maquina_final = dados_maquina.copy()

    puntos_jugador = dados.puntuacion(dados_jugador_final)
    puntos_maquina = dados.puntuacion(dados_maquina_final)

    print("\n--- RESULTADOS ---")
    print(f"Tú sacaste: {dados_jugador_final} -> Total: {puntos_jugador}")
    print(f"Yo saqué: {dados_maquina_final} -> Total: {puntos_maquina}")
    print("------------------")

    if puntos_jugador > puntos_maquina:
        print("¡Me ganaste! Muy bien jugado.")
    elif puntos_jugador < puntos_maquina:
        print("¡Gané yo! Mejor suerte la próxima.")
    else:
        print("¡Empatamos! Qué casualidad.")
````

## Ejercicio 2: módulo clase Frecuencias

**Descripción:**  
Queremos definir el módulo **frecuencias.py** que contiene una clase Frecuencias que permita crear objetos que calculen la frecuencia de los **símbolos alfabéticos** que aparezcan en cadenas que se le proporcionen. 
Se utilizará un diccionario para representar la tabla de frecuencias. Las claves del diccionario serán los símbolos y los valores un entero positivo con el número de apariciones.
Por ejemplo, `"a":5` significa que la `a` aparece 5 veces desde que se creó el objeto de la clase Frecuencias.

Recuerda que para calcular las frecuencias no tiene sentido distinguir entre mayúsculas y minúsculas.


**Requisitos:**

- **Atributos:**
  - `__cadena`: cadena de texto que se ha leido hasta el momento (atributo privado).
  - `__frecuencias`: diccionario vacío inicialmente (atributo privado).

- **Métodos:**
  - `__init__(self)`
    - Constructor que inicializa la cadena y el diccionario 'frecuencias` vacío.
  - `add_cadena(self, nueva_cadena)`
    - Añade 'nueva_cadena' a la cadena almacenada e incrementa las frecuencias absolutas de cada símbolo que aprece en 'nueva_cadena'.
  - `add_fichero(self, nombre_fichero)`
    - Añade el contenido del fichero de texto 'nombre_fichero' a la cadena almacenada e incrementa las frecuencias absolutas de cada símbolo que aprece en 'nueva_cadena'.
  - `get_cadena(self)`
    - Devuelve la cadena almacenada hasta este momento.
  - `print_fichero(self,nombre_fichero)`
    - Escribe la cadena almacenada hasta este momento en el fichero de texto de nombre 'nombre_fichero'
  - `frecuencia_relativa(self,simbolo)`
    - Devuelve la frecuencia en % del símbolo proporcionado, es decir su relación con respecto al número total de símbolos alfabéticos almacenados.
  - `__str__(self)`
    - Devuelve una representación legible del diccionario de frecuencias. Para cada letra guardada en el diccionario se quiere imprimir una línea de la forma "La letra {letra} ha aparecido {num_veces} veces." Las letras han de aparecer en orden alfabético.
