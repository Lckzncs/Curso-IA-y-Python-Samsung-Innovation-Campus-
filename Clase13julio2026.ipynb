# Elige al azar 23 números entre el 2 y el 4. Calcula su suma e imprímela por pantalla.


```python
import random

suma = 0

print("Los números aleatorios son:")

for i in range(23):
    n = round(2 + random.random() * 2, 2)
    # Random.random() genera un número entre 0 y 1, si lo multiplicamos por 2 daría un número entre 0 y 2,
    # y al sumarle 2 ya nos aseguramos de que empezamos a partir de dos.
    print(n)
    suma += n

print(f"La suma es: {suma}")
```

# Elige 100 números impares al azar entre 1 y 10001. Cuenta cuántos de ellos son primos, e imprímelo por pantalla.

```python
import random
import math
#Definimos una función que nos devuelva un booleano True si el número es primo y False si no lo es.
def es_primo(n):
    primo = True
    if n ==1:
        primo = False
    num = 2
    while primo and num <= math.sqrt(n):
        if n % num == 0:
            primo = False
        num += 1
    return primo

num_primos = 0
print("Los números primos entre los 100 números entre 1 y 10001 escogidos al azar son:", end = " ")
for idx in range(100):
    n = random.randrange(1,10002,2)
    if es_primo(n):
        num_primos +=1
        print(n,end = " ")
print()
print(f"En total, había: {num_primos} primos.")
```

```python
import random

numeros = []
for i in range(100):
    numeros.append(random.randrange(1, 10002, 2))

primos = 0
for n in numeros:
    contador = 0
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            contador += 1
    if n >= 2 and contador == 0:
        primos += 1

print("Números elegidos:", numeros)
print("Cantidad de primos:", primos)
```

# Let's code

Vamos a programar un juego similar a una máquina tragaperras. El usuario humano juega contra el ordenador. La máquina tragaperras tiene un número de casillas que se especifica al principio del programa. El humano puede darle a un boton tantas veces como quiera. Cada vez que pulsa el botón, el contenido de todas las casillas cambia. Cada casilla contiene el dibujo de una fruta. Hay tres tipos distintos de frutas. Cuando el humano decide dejar de darle al botón, se calcula su puntuación. La puntuación es el número de veces que aparece la fruta más repetida. El ordenador realiza el mismo proceso, y se calcula su puntuación. El ganador es el que consigue mayor puntuación.
El programa debe imprimir por pantalla todo el proceso, incluyendo las distintas tiradas. Los dibujos de las frutas pueden ser caracteres Unicode.

```python
import random

CASILLAS = 5
FRUTAS = ["🍎", "🍋", "🍇"]

print("============================")
print("   🎰 MÁQUINA TRAGAPERRAS   ")
print("============================")
print(f"Casillas: {CASILLAS}  |  Frutas: {' '.join(FRUTAS)}")
print()

print(">>> TU TURNO <<<")
tiradas_humano = []

entrada = input("Pulsa ENTER para girar o escribe 'stop' para parar: ")
while entrada.lower() != "stop":
    casillas = [random.choice(FRUTAS) for _ in range(CASILLAS)]
    tiradas_humano.append(casillas)
    print(f"  Tirada {len(tiradas_humano)}: {' '.join(casillas)}")
    entrada = input("Pulsa ENTER para girar o escribe 'stop' para parar: ")

if tiradas_humano:
    ultima_humano = tiradas_humano[-1]
    puntos_humano = max(ultima_humano.count(f) for f in FRUTAS)
else:
    ultima_humano = []
    puntos_humano = 0

print(f"\nTu última tirada: {' '.join(ultima_humano) if ultima_humano else 'Sin tiradas'}")
print(f"Tu puntuación: {puntos_humano} puntos")

print("\n>>> TURNO DEL ORDENADOR <<<")
num_tiradas = random.randint(1, 6)
tiradas_pc = []

for i in range(num_tiradas):
    casillas = [random.choice(FRUTAS) for _ in range(CASILLAS)]
    tiradas_pc.append(casillas)
    print(f"  Tirada {i + 1}: {' '.join(casillas)}")

ultima_pc = tiradas_pc[-1]
puntos_pc = max(ultima_pc.count(f) for f in FRUTAS)

print(f"\nÚltima tirada del ordenador: {' '.join(ultima_pc)}")
print(f"Puntuación del ordenador: {puntos_pc} puntos")

print()
print("============================")
print("      RESULTADO FINAL       ")
print("============================")
print(f"  Tu puntuación:            {puntos_humano}")
print(f"  Puntuación del ordenador: {puntos_pc}")
print()

if puntos_humano > puntos_pc:
    print("🎉 ¡Ganaste!")
elif puntos_pc > puntos_humano:
    print("😞 Ganó el ordenador.")
else:
    print("🤝 ¡Empate!")
```

```python
import random
def tirada(num_casillas, dibujos):
    lista_casillas = []
    for i in range(num_casillas):
        lista_casillas.append(random.choice(dibujos))
    return lista_casillas
def puntuacion(tirada, dibujos):
    maxi = 0
    for dibu in dibujos:
        repes = 0
        for eleccion in tirada:
             if dibu == eleccion:
                 repes += 1
        if repes >= maxi:
            maxi = repes
    return maxi

#dibujos = ["pera", "plátano", "naranja"]
dibujos =[ "\U0001F350","\U0001F34C","\U0001F34A"]

num_casillas=int(input("Introduzca el número de casillas que tendrá el juego:"))
nombre = input("Introduce tu nombre para comenzar para comenzar:")
comienzo = nombre
while comienzo != "":
    tirada_jugador = tirada(num_casillas, dibujos)
    print("Tu tirada ha sido:",tirada_jugador)
    comienzo = input("Si quieres generar una nueva tirada pulsa cualquier botón, si no, pulsa Enter:")
repetir = True
while repetir:
    tirada_maquina = tirada(num_casillas, dibujos)
    puntuacion_maquina = puntuacion(tirada_maquina, dibujos)
    if puntuacion_maquina > num_casillas // 2:
        repetir = False
puntuacion_jugador = puntuacion(tirada_jugador, dibujos)
print(f"La máquina ha obtenido una puntuación de: {puntuacion_maquina} con la tirada {tirada_maquina}")
print(f"{nombre} ha obtenido una puntuación de: {puntuacion_jugador} con la tirada {tirada_jugador}")
if puntuacion_maquina > puntuacion_jugador:
    print("Ganó la máquina!!")
elif puntuacion_maquina < puntuacion_jugador:
    print(f"Ganó  {nombre} !!")
else:
    print("Empate!!")
```
