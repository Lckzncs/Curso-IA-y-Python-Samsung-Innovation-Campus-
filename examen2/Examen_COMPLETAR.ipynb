## Apartado 1: Clase Base (2puntos)

Diseña la clase `Tripulante`, que vendrá representada por su `nombre`, `edad` y `codigo_identificador`. Estos atributos serán privados. La clase ofrecerá el siguiente constructor y métodos públicos:

* `def __init__(self, codigo_identificador, nombre, edad)`: Construye un objeto `Tripulante`. Recibe como parámetros, en ese orden, el código identificador, el nombre y su edad. Si el nombre o el código identificador son cadenas vacías `""`, o la edad es menor de `18` se debe lanzar una excepción de clase `ValueError`
  
*`def nombre(self)`, `def edad(self)` y `def codigo_identificador(self)`: Son métodos para obtener el valor del atributo respectivo. Se debe utilizar decoradores para definirlos.

* `def __str__(self)`: Devuelve una representación en forma de cadena del tripulante, según el formato del siguiente ejemplo:
  > `(NX-01, Jonathan Archer, 45 años)`
  
* `def __eq__(self, other)`: Compara el tripulante actual (`self`) con el tripulante recibido como parámetro (`other`). Dos tripulantes serán iguales sólo si su código identificador es el mismo, independientemente de mayúsculas y minúsculas.

```python
# SOLUCIÓN APARTADO 1
class Tripulante:
    def __init__(self, codigo_identificador, nombre, edad):
        if nombre == "" or codigo_identificador == "" or edad < 18:
            raise ValueError("El nombre o código no pueden estar vacíos, y la edad debe ser al menos 18.")
        
        self._codigo_identificador = codigo_identificador
        self._nombre = nombre
        self._edad = edad

    @property
    def nombre(self):
        return self._nombre

    @property
    def edad(self):
        return self._edad

    @property
    def codigo_identificador(self):
        return self._codigo_identificador

    def __str__(self):
        return f"({self.codigo_identificador}, {self.nombre}, {self.edad} años)"

    def __eq__(self, other):
        if isinstance(other, Tripulante):
            return self.codigo_identificador.lower() == other.codigo_identificador.lower()
        return False
```

## Apartado 2: Clase Oficial (2.5 puntos)

La clase `Oficial` hereda de la clase `Tripulante`. En ella se permite especificar la `division` científica o militar a la que pertenece (por ejemplo: 'Mando', 'Ciencia', 'Ingeniería'), que será un atributo privado de esta clase derivada. La clase ofrecerá el siguiente constructor y métodos públicos:

* `def __init__(self, codigo_identificador, nombre, edad, division)`: Construye un objeto `Oficial`. Recibe como parámetros, en ese orden, el código, el nombre, la edad y la división.
* `def division(self)`: Devuelve la información de la división del oficial. se deben usar decoradores.
* `def __str__(self)`: Devuelve la representación textual del objeto, según el formato del siguiente ejemplo:
  > `[División: Mando] - (NX-01, Jonathan Archer, 45 años)`
* `def calcular_puntos_mision(self, simulaciones)`: **Método Recursivo** que calcula los puntos de experiencia acumulados para la próxima misión espacial basándose en el número de simulaciones completadas. 
  * Si no se han realizado simulaciones, se parte de un valor de `10` puntos.
  * Si se han realizado simulaciones, la puntuación se calcula como `(simulaciones * 5) + calcular_puntos_mision(simulaciones - 1)`.

```python
# SOLUCIÓN APARTADO 2
class Oficial(Tripulante):
    def __init__(self, codigo_identificador, nombre, edad, division):
        Tripulante.__init__(self, codigo_identificador, nombre, edad)
        self._division = division

    @property
    def division(self):
        return self._division

    def __str__(self):
        return f"[División: {self.division}] - {Tripulante.__str__(self)}"

    def calcular_puntos_mision(self, simulaciones):
        if simulaciones == 0:
            return 10
        
        return (simulaciones * 5) + self.calcular_puntos_mision(simulaciones - 1)
```

## Apartado 4: Clase NaveEspacial (4 puntos)

Defina la clase `NaveEspacial`, que representará una nave de exploración. Esta clase contendrá de manera privada los siguientes atributos: una lista de oficiales (`oficiales`) y una lista que funcionará bajo la estructura de **Cola (FIFO)** para almacenar misiones espaciales pendientes (`cola_misiones`).

Ofrecerá el constructor y los métodos descritos a continuación:

* `def __init__(self)`: Construye un objeto `NaveEspacial` vacío (sin oficiales y con la cola de misiones vacía).
  
* `def agregar_oficial(self, codigo, nombre, edad, division)`: Crea un nuevo objeto `Oficial` con los parámetros recibidos. Si el oficial ya existe en la lista de la nave (utilizando la igualdad definida previamente), mostrará un mensaje de aviso por pantalla; en caso contrario, lo añadirá a la lista.

* `def __str__(self)`: Devuelve la representación textual de la nave mostrando entre llaves `{}` todos los oficiales pertenecientes a la flota (separados por un salto de línea).
  Ejemplo:
  >{
  >
  >[División: Mando] - (NCC-1701-A, James T. Kirk, 34 años)
  >
  >[División: Ingeniería] - (NCC-1701-B, Montgomery Scott, 42 años)
  >
  >[División: Médica] - (NCC-1701-C, Leonard McCoy, 42 años)
  >
  >[División: Ingeniería] - (NCC-1701-D, Geordi La Forge, 29 años)
  >
  >}
  
* `def obtener_diccionario_edades(self)`: Devuelve un diccionario que tiene como claves un número que representa la edad de los oficiales y como valor a asociado a cada edad una lista con los codigos de identificación de los oficiales que tienen esa edad.

* `def guardar_nave(self, nombre_archivo)`: Recibe el nombre del archivo y guarda la lista completa de oficiales en un archivo binario utilizando el módulo `pickle`.

* `def obtener_divisiones_unicas(self)`: Utiliza **Conjuntos (Sets)** para procesar la lista de oficiales y devolver un conjunto con los nombres únicos de las divisiones presentes en la nave.

* `def encolar_mision(self, codigo_mision)`: Añade un código de misión (string) al final de la cola de misiones pendientes de la nave.

* `def desencolar_mision(self)`: Extrae y devuelve la misión que se encuentra al principio de la cola. Si la cola está vacía, debe retornar un mensaje indicándolo.

```python
# SOLUCIÓN APARTADO 4
class NaveEspacial:
    def __init__(self):
        self._oficiales = []
        self._cola_misiones = []

    def agregar_oficial(self, codigo, nombre, edad, division):
        nuevo_oficial = Oficial(codigo, nombre, edad, division)
        
        if nuevo_oficial in self._oficiales:
            print("Aviso: El oficial ya se encuentra registrado en la nave.")
        else:
            self._oficiales.append(nuevo_oficial)

    def __str__(self):
        representacion = "{\n"
        for oficial in self._oficiales:
            representacion += f"{oficial}\n"
        representacion += "}"
        return representacion

    def obtener_diccionario_edades(self):
        edades = {}
        for oficial in self._oficiales:
            if oficial.edad not in edades:
                edades[oficial.edad] = []
            edades[oficial.edad].append(oficial.codigo_identificador)
        return edades

    def guardar_nave(self, nombre_archivo):
        with open(nombre_archivo, 'w', encoding='utf-8') as archivo:
            for oficial in self._oficiales:
                archivo.write(f"{oficial.codigo_identificador},{oficial.nombre},{oficial.edad},{oficial.division}\n")

    def obtener_divisiones_unicas(self):
        return {oficial.division for oficial in self._oficiales}

    def encolar_mision(self, codigo_mision):
        self._cola_misiones.append(codigo_mision)

    def desencolar_mision(self):
        if len(self._cola_misiones) == 0:
            return "La cola de misiones está vacía."
        
        return self._cola_misiones.pop(0)
```
