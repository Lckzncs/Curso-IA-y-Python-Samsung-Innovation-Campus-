# Operadores relacionales o de comparación

Los operadores relacionales permiten comparar dos valores, como si se tratara de una balanza.



Por ejemplo, con el operador relacional `==` podemos comprobar si dos literales numéricos son iguales:

````python
5 == 5
````

````python
6 == 8
````

Observa que Python devuelve dos resultados: `True` y `False`.

Podemos preguntar a Python de qué tipo son `True` y `False`:

````python
print(type(True))
print(type(False))
````

El tipo `bool` es un tipo lógico que solo tiene dos valores: cierto y falso. Como tiene dos valores, solo tiene dos **literales**, que son `True` (cierto) y `False` (falso), respectivamente.

Hasta ahora conocemos cuatro tipos, cada uno con sus literales:

| Tipo    | Literales         |
|---------|-------------------|
| `int`   |  `5`, `12`,...    |
| `float` | `3.14`, `6.0`,... |
| `str`   | `"python\n"`,...  |
| `bool`  | `True`, `False`   |

Comparado con los otros tipos que conocemos (`int`, `float`, `str`), el tipo `bool` parece muy simple. Para los tipos `int`, `float` y `str` tenemos _"infinitos"_ valores (y por tanto, _"infinitos"_ literales). Sin embargo, para el tipo `bool` solo tenemos dos valores y, por lo tanto, dos literales. Sin embargo, el tipo `bool` es muy potente. Es muy importante conocerlo a fondo para programar correctamente.

La comparación de igualdad no está limitada a literales enteros; podemos comprobar si dos expresiones aritméticas son iguales (es decir, si su evaluación produce el mismo valor):

````python
5 + 3 == 9 
````

Obviamente, las expresiones aritméticas pueden incluir variables:

````python
7- 4 == 1 + 2
````

````python
peso = 57
print(peso + 1 == 58)
````

Observa que las celdas anteriores evalúan una expresión que devuelve un resultado de tipo `bool`:
```python
    expresion_aritmetica_1 == expresión_aritmética_2
```
A las expresiones que devuelven un resultado de tipo `bool` se les llama **expresiones booleanas**. Ya conocemos tres tipos de expresiones: aritméticas, de cadena y booleanas.

````python
1 + 6 + 12 #expresión aritmetica
````

````python
7 * 2 == 10 + 4 # expresión logica o booleana
````

````python
"Hola" + " " + "mundo" # expresión de cadena de caracteres
````

### Expresiones y sentencias

Recuerda que no es lo mismo una expresión que una sentencia:

> - una expresión se evalúa y devuelve un valor
> - una sentencia se ejecuta y tiene un efecto

Aunque `=` y `==` se parecen mucho, son muy diferentes. El primero (`=`) es una sentencia; el segundo (`==`) es un operador relacional que aparecerá en expresiones booleanas.

Observa la diferencia:

````python
a = 2
print (a)
````

````python
a == 2
````

````python
a == 3
````

Además del operador relacional `==`, Python cuenta con operadores relacionales para realizar todo tipo de comparaciones:

| Operador | Significado       |
|----------|-------------------|
|  `==`    | igual que         |
|  `!=`    | distinto que      |
|  `<`     | menor que         |
|  `>`     | mayor que         |
|  `<=`    | menor o igual que |
|  `>=`    | mayor o igual que |

Vamos a comparar algunas expresiones aritméticas (de tipos `int` o `float`) para comprobar su funcionamiento.

````python
5 == 7
````

````python
5 <= 7
````

````python
5 != 7
````

````python
5 >= 7
````

````python
nota = 7
print(5 <= nota <= 10)
print((5 <= nota) and (nota <= 10))
````

## Comparación de cadenas

Hasta ahora hemos utilizado los operadores relacionales para comparar valores numéricos. En realidad, Python es capaz también de comparar cadenas.

- Las cadenas se comparan carácter a carácter, del primero al último
- Los caracteres no están ordenados como en el diccionario; por ejemplo:
   - los dígitos son menores que las letras
   - las letras mayúsculas son menores que las minúsculas
   - las letras con acentos, etc. son mayores que el resto de letras

Estas reglas de ordenación pueden dar lugar a algunas sorpresas.

````python
"3" < "A"
````

````python
"a" < "A"
````

````python
print(3 != "A")
print(3 == "A")
````

````python
"casa" < "caserio"
````

````python
"Angeles" < "Maria"
````

````python
x1 = "Casa"
x2 = "casa"
x1 == x2
````

## Errores de tipo con los operadores relacionales

Los operadores relacionales pueden comparar dos valores numéricos o dos cadenas; pero no pueden comparar un valor numérico y una cadena:

````python
 3 <"3"
````

Sin embargo, Python puede realizar conversiones de tipo implícitas en los operandos; por ejemplo, los enteros se convierten de forma implícita a flotantes:

````python
3 < 3.14
````

Se pueden aplicar los operadores `==` y `!=` a cualquier par de valores, independientemente de su tipo. Aunque no produce un error, **no es buena idea**. Es mejor comparar solo valores del mismo tipo:

````python
3 != "A"
````

````python
3 == "3"
````

## Tipo booleano (`bool`)

De la misma forma que hemos utilizado variables de tipo `int` o `float`, podemos crear variables de tipo `bool`. A simple vista esto no parece muy útil, pero veremos que resultan muy útiles. Por ejemplo, podemos introducir una variable que indique si una persona está jubilada:

````python
esta_jubilada = True
````

````python
esta_jubilada
````

````python
esta_jubilada = "si" # No es la mejor opción
````

A las funciones de conversión que conocíamos (`int()`, `float()` y `str()`) les añadimos ahora la función `bool()`, que permite convertir cualquier expresión en un valor booleano (`True` o `False`):

````python
print(int(5.3))
print(bool(5.3))
````

````python
print(bool(-4.2))
````

````python
print(bool(0))
````

````python
print(bool(0))
print(bool(0.0))
print(bool(0.00000000000))
````

````python
print(bool("Python"))
print(bool("      "))
print(bool(""))
print(bool(None))
````

De los ejemplos anteriores concluimos que al convertir un valor de cierto tipo a `bool`:
> - si el tipo es numérico, si valor es cero se convierte `False`, sino `True`
> - si el tipo _tiene cosas dentro_ (como las cadenas), si el valor está está vacío es `False`, sino `True`

## Operadores lógicos o booleanos

Para los tipos `int`, `float` y `str` disponemos de varios operadores que nos permiten escribir expresiones:

````python
32 * 6 // (7 - 3)
````

````python
2 ** 4.1 * 1.45
````

````python
"hola" + "adios" [-1] * 10
````

Los operadores relacionales devuelven un resultado de tipo `bool`:

````python
5 <= 10
````

Además, con el tipo `bool` podemos usar los operadores booleanos o lógicos:

| Operador | Significado |
|----------|-------------|
| `and`    | conjunción  |
| `or`     | disyunción  |
| `not`    | negación    |

Estos operadores toman operandos de tipo `bool` y devuelven `bool`como resultado. Su funcionamiento se suele ilustrar mediante _tablas de verdad_.

> `a and b`: es cierto si `a` y `b` son ciertos.

|  `a`  |  `b`  | `a and b` |
|:-----:|:-----:|:---------:|
| False | False | False     |
| False | True  | False     |
| True  | False | False     |
| True  | True  | True      |

> `a or b`: es cierto si `a` o `b` son ciertos.

|  `a`  |  `b`  |  `a or b` |
|:-----:|:-----:|:---------:|
| False | False | False     |
| False | True  | True      |
| True  | False | True      |
| True  | True  | True      |

> `not a`: es cierto si `a` es falso y viceversa.

|  `a`  | `not a` |
|:-----:|:-------:|
| False | True    |
| True  | False   |



````python
(5 < 7) and ("a" in "hola")
````

````python
(5 < 7) and ("e" in "hola")
````

````python
(5 < 7) or ("e" in "hola")
````

````python
not (5 < 7)
````

````python
not (5 < 7) or ("e" in "hola")
````

````python
5 < 7 and 8 < 10 and 100 < 122 and 45 < 234
````

````python
(5 < 7) and (8 < 10) and (100 < 122) and (45 < 234)
````

## Prioridad o precedencia de operadores

A los operadores aritméticos y de cadena, hemos añadido los operadores relacionales y los lógicos. Todos estos operadores pueden combinarse en una sola expresión. La prioridad o precedencia de los operadores está definida de manera que haya que poner pocos paréntesis:

- primero se aplican los operadores numéricos
- después se aplican los operadores relacionales
- finalmente se aplican los operadores lógicos

La siguiente tabla muestra los operadores Python que hemos estudiado ordenados por prioridad:

| Operador                       | Descripción               |
|--------------------------------|---------------------------|
| ()                             | paréntesis                |
| **                             | potencia                  |
| + -                            | unarios                   |
| * / // %                       | multiplicativos           |
| + -                            | aditivos                  |
|  == != > >= < <= `in` `not in` | relacionales, pertenencia |
| `not`                          | negación                  |
| `and`                          | conjunción                |
| `or`                           | disyunción                |


Los operadores relacionales y lógicos nos permiten escribir **condiciones** en Python. Es muy importante aprender a usar bien estos operadores, pues las condiciones son fundamentales en la programación.

11. Un fábrica cuenta con un sofisticado sistema de seguridad industrial. La alerta se debe activar si se cumple alguna de las siguientes condiciones:

- La temperatura es mayor que 95.0 y el el sensor está activado.
- La presión es menor que 10.0 o mayor que 1000.0, y el sensor está activado.
- El sensor está desactivado y la temperatura es mayor que 80.0 pero la presión es menor o igual a 500.0.
- La temperatura es exactamente 0.0 o la presión es exactamente 0.0, y el sensor está desactivado.

````python
temperatura = 90.0
presion = 1010.0
sensor_activado = True

condicion_1 = (temperatura > 95.0) and sensor_activado
condicion_2 = ((presion < 10.0) or (presion > 1000.0)) and sensor_activado
condicion_3 = (not sensor_activado) and (temperatura > 80.0) and (presion <= 500.0)
condicion_4 = ((temperatura == 0.0) or (presion == 0.0)) and (not sensor_activado)

alerta = condicion_1 or condicion_2 or condicion_3 or condicion_4

print("¿La alerta se debe activar?:", alerta)
````

## Operadores lógicos y evaluación por cortocircuito

Los operadores lógicos `and` y `or` solo evalúan el segundo operando cuando es necesario. A esto se le llama evaluación por cortocircuito:

````python
(5 > 7) and (6 // 2 == 3)
````

````python
(6 // 0 == 3)
````

La evaluación por cortocircuito es especialmente importante cuando se combina con la indexación:

````python
(5 < 7) or (6 // 0 == 0)
````
