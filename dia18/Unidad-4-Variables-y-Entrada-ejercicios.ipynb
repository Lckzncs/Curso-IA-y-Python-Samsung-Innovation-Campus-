## Ejercicio 16

Escribe un programa que lea por teclado el **diámetro** de una esfera (en cm) y calcule su **volumen** (en cm³). Usa `3.1416` como valor de pi.

$$V = \frac{4}{3}\, \pi r^3$$

Recuerda que el radio es la mitad del diámetro.

````python
diametro = float(input("Introduce el diámetro de la esfera en cm: "))

radio = diametro / 2
volumen = (4 / 3) * 3.1416 * (radio ** 3)

print("El volumen de la esfera es:", volumen, "cm³")
````

## Ejercicio 18

Escribe un programa que lea por teclado una cantidad de segundos (por ejemplo `7384`) y la muestre descompuesta en **horas**, **minutos** y **segundos**:

```text
7384 segundos son 2 horas, 3 minutos y 4 segundos
```

Pista: usa división entera (`//`) y módulo (`%`).

````python
segundos_totales = int(input("Introduce la cantidad de segundos: "))

horas = segundos_totales // 3600
segundos_restantes = segundos_totales % 3600
minutos = segundos_restantes // 60
segundos = segundos_restantes % 60

print(segundos_totales, "segundos son", horas, "horas,", minutos, "minutos y", segundos, "segundos")
````

## Ejercicio 19

Una cafetería tiene una máquina expendedora que solo devuelve cambio en monedas de **2 €**, **1 €** y **0,20 €**. Escribe un programa que pida el dinero introducido y el precio del producto (ambos en céntimos, para trabajar con enteros) y muestre cuántas monedas de cada tipo hay que devolver, usando el menor número posible.

```text
¿cuánto dinero pones en la máquina (en céntimos)? 500
¿cuánto vale el producto (en céntimos)? 140

El cambio es:
monedas de 200 céntimos= 1
monedas de 100 céntimos= 1
monedas de 20 céntimos= 3
```

````python
dinero_introducido = int(input("¿cuánto dinero pones en la máquina (en céntimos)? "))
precio_producto = int(input("¿cuánto vale el producto (en céntimos)? "))

cambio_total = dinero_introducido - precio_producto

monedas_200 = cambio_total // 200
resto = cambio_total % 200

monedas_100 = resto // 100
resto = resto % 100

monedas_20 = resto // 20

print("\nEl cambio es:")
print("monedas de 200 céntimos =", monedas_200)
print("monedas de 100 céntimos =", monedas_100)
print("monedas de 20 céntimos =", monedas_20)
````

## Ejercicio 20

Escribe un programa que lea por teclado la **altura** $a$ y la **base** $b$ de un triángulo rectángulo y calcule, en este orden:

1. su **hipotenusa** $c$ (teorema de Pitágoras)
2. su **perímetro** $p = a + b + c$
3. su **área** $A = \dfrac{b \cdot a}{2}$

Muestra los tres resultados con `print()`.

Recuerda que:

$$c = \sqrt{a^2 + b^2}$$

Puedes calcular la raíz cuadrada elevando a `0.5` o usando `math.sqrt()`.

````python
base = float(input("Introduce la base del triángulo rectángulo: "))
altura = float(input("Introduce la altura del triángulo rectángulo: "))

hipotenusa = (base ** 2 + altura ** 2) ** 0.5
perimetro = base + altura + hipotenusa
area = (base * altura) / 2

print("Hipotenusa:", hipotenusa)
print("Perímetro:", perimetro)
print("Área:", area)
````
