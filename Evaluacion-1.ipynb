### Pintando rectángulos (1.5 puntos)

Escribe un programa que lea la altura y anchura de un rectángulo por teclado e imprima un rectángulo de asteriscos (`*`) con dichas dimensiones. Puedes suponer que las dimensiones facilitadas serán siempre enteros positivos mayores o iguales que 3.  

```
dame la altura (>=3): 4
dame al anchura (>=3): 10
**********
*        *
*        *
**********

dame la altura (>=3): 3
dame al anchura (>=3): 3
***
* *
***

```

````python
altura = int(input("dame la altura (>=3): "))
anchura = int(input("dame la anchura (>=3): "))

for linha in range(altura):

    if linha == 0 or linha == altura - 1:
        print("*" * anchura)
    else:
        print("*" + " " * (anchura - 2) + "*")
````

### Sumando  positivos y negativos (1.5 puntos)

Escribe un programa que lea números enteros por teclado hasta que se introduzca un cero. El programa debe mostrar entonces dos sumas: la suma de todos los números positivos introducidos y la suma de todos los negativos. 

```
dame un entero (0 para acabar): 1
dame un entero (0 para acabar): -2
dame un entero (0 para acabar): 3
dame un entero (0 para acabar): -4
dame un entero (0 para acabar): -5
dame un entero (0 para acabar): 0
La suma de los positivos es: 4
La suma de los negativos es: -11
```

````python
suma_positivos = 0
suma_negativos = 0

numero = int(input("dame un entero (0 para acabar): "))

while numero != 0:
    
    if numero > 0:
        suma_positivos += numero
        
    elif numero < 0:
        suma_negativos += numero
        
    numero = int(input("dame un entero (0 para acabar): "))

print(f"La suma de los positivos es: {suma_positivos}")
print(f"La suma de los negativos es: {suma_negativos}")
````

### Números de Armstrong (1.5 puntos)

Un número de Armstrong es un entero de tres cifras $n$ tal que la suma de los cubos de sus cifras es igual a $n$. Por ejemplo, $153$ es un número de Armstrong, porque $1^3 + 5^3 + 3^3 = 153$.

Escribe un programa que imprima todos los números de Armstrong.

````python
for n in range(100, 1000):
    centenas = n // 100
    decenas = (n // 10) % 10
    unidades = n % 10
    
    suma = (centenas**3) + (decenas**3) + (unidades**3)
    
    if suma == n:
        print(n)
````

### Piedra, papel, tijera por rondas (3 puntos)

Escribe un programa que lea por teclado cuántas rondas van a jugar dos jugadores a _piedra, papel, tijera_. Después, para cada ronda, debe leer la jugada del primer jugador y la jugada del segundo jugador. Las jugadas introducidas deben ser siempre `piedra`, `papel` o `tijera`, en caso contrario debe pedir la jugada de nuevo.

El programa debe imprimir quién gana cada ronda, llevar la cuenta de las victorias de cada jugador y de los empates, y mostrar el resultado final al terminar todas las rondas. Debes resolverlo usando un bucle.

Recuerda que:

- la piedra vence a la tijera y pierde frente al papel
- el papel vence a la piedra y pierde frente a la tijera
- la tijera vence al papel y pierde frente a la piedra

```
¿Cuántas rondas se van a jugar?: 3
Ronda 1
jugador 1: piedra
jugador 2: papel
Ha ganado el segundo jugador
Ronda 2
jugador 1: tijera
jugador 2: papel
Ha ganado el primer jugador
Ronda 3
jugador 1: piedra
jugador 2: piedra
Han empatado
Victorias del primer jugador: 1
Victorias del segundo jugador: 1
Empates: 1
Resultado final: empate
```

````python
rondas = int(input("¿Cuántas rondas se van a jugar?: "))
puntos_j1 = 0
puntos_j2 = 0
empates = 0

for ronda in range(rondas):
    print(f"Ronda {ronda + 1}")
    
    j1 = input("jugador 1: ")
    while j1 != "piedra" and j1 != "papel" and j1 != "tijera":
        j1 = input("jugador 1: ")
        
    j2 = input("jugador 2: ")
    while j2 != "piedra" and j2 != "papel" and j2 != "tijera":
        j2 = input("jugador 2: ")
        
    if j1 == j2:
        print("Han empatado")
        empates += 1
    elif (j1 == "piedra" and j2 == "tijera") or (j1 == "papel" and j2 == "piedra") or (j1 == "tijera" and j2 == "papel"):
        print("Ha ganado el primer jugador")
        puntos_j1 += 1
    else:
        print("Ha ganado el segundo jugador")
        puntos_j2 += 1

print(f"Victorias del primer jugador: {puntos_j1}")
print(f"Victorias del segundo jugador: {puntos_j2}")
print(f"Empates: {empates}")

if puntos_j1 > puntos_j2:
    print("Resultado final: El primer jugador es el campeón!!!")
elif puntos_j2 > puntos_j1:
    print("Resultado final: El segundo jugador es el campeón!!!")
else:
    print("Resultado final: empate")
````

### Contando vocales (1 punto)

Escribe un programa que lea una frase en minúsculas y sin tildes, y cuente cuántas vocales contiene. Deben considerarse vocales los caracteres `a`, `e`, `i`, `o` y `u`.

El programa debe recorrer la cadena usando un bucle y mostrar el número total de vocales encontradas.

```
dame una frase en minúsculas y sin tildes: hola mundo
Número de vocales: 4

dame una frase en minúsculas y sin tildes: python
Número de vocales: 1
```

````python
frase = input("dame una frase en minúsculas y sin tildes: ")

contador = 0

for letra in frase:
    
    if letra in "aeiou":
        contador += 1

print(f"Número de vocales: {contador}")
````
