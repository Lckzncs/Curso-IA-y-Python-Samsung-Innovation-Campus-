# Unidad 6 - Sentencias Condicionales - I

Hasta el momento todos nuestros programas Python han utilizado la **estructura de control** más simple; la **secuencia**:



La secuencia se caracteriza porque todos los pasos que la componen se ejecutan uno tras otro, exactamente en el orden descrito.

Por ejemplo, el programa que calculaba el cambio de una máquina expendedora (en _wons_) utiliza la secuencia:

```python
pago = int(input("¿cuánto dinero pones en la máquina?"))
precio = int(input("¿cuánto vale el producto?"))

cambio = pago - precio
monedas_de_500 = cambio // 500
cambio_pendiente = cambio % 500
monedas_de_100 = cambio_pendiente // 100

print("El cambio es:")
print("monedas de 500 won=", monedas_de_500)
print("monedas de 100 won=", monedas_de_100)
```

Puedes pensar en la secuencia como en una carretera que discurre en línea recta: es fácil conducir, pero resulta aburrido.



Además de aburrida, la secuencia no siempre resulta adecuada: a veces hay algunos pasos que nos gustaría saltarnos.

Por ejemplo, el siguiente programa calcula la velocidad a la que nos hemos desplazado:

```python
distancia = float(input("¿Qué distancia en km has recorrido? "))
tiempo = float(input("¿Cuánto tiempo en horas has invertido? "))

velocidad = distancia / tiempo

print("Tu velocidad ha sido de", velocidad, "Km/h")
```

¿Qué pasa si en el anterior programa el tiempo invertido es `0.0`? El cálculo de la velocidad falla con división por cero. Nos gustaría saltarnos ese paso si la velocidad es `0.0`.

Es frecuente que algunas sentencias solo se deben ejecutar si se satisfacen ciertas **condiciones**, por ejemplo:
- se calcula la velocidad si el tiempo invertido no es 0
- la máquina expendedora no permite completar la compra porque no se ha introducido suficiente dinero
- se aplica descuento de carnet joven si tienes menos de 20 años
- has conseguido el objetivo si has andado al menos 10000 pasos

Siguiendo con el símil de la carretera, necesitamos indicar de alguna forma que nuestro programa puede tomar más de un camino; que existe una bifurcación:



## La selección 

Para especificar estas situaciones necesitamos utilizar una nueva estructura de control: la **selección**.

La selección permite indicar que un paso es **condicional**; el paso solo se ejecuta si cierta condición (a veces llamada _guarda_) es cierta.

Si la condición es cierta, el paso se ejecuta, en caso contrario se salta:



A esta estructura de control se le llama **condicional simple**. Más adelante introduciremos otros tipos de estructuras de control selectivas más potentes.

## La condicional simple en Python

Para expresar una condicional simple en Python usamos una sentencia `if` con la siguiente estructura:



La parte condicional que aparece tras los dos puntos `:` es un **bloque** y puede estar formada por más de una sentencia:
- todas las sentencias que forman el bloque deben estar sangradas al mismo nivel
- la recomendación de estilo PEP 8 es que los bloques se sangren 4 espacios
- al sangrar debemos utilizar espacios, nunca tabuladores
- el bloque termina cuando aparece una sentencia que está sangrada a un nivel igual o inferior al `if`

Con la sentencia `if`, podemos evitar que el programa que calcula la velocidad falle si el tiempo invertido es 0.

**Recuerda:** los bloques deben estar convenientemente sangrados, de lo contrario Python protesta.

**Ejercicio:** Escribe un programa Python que lea el número de pasos que se han dado y que indique si se ha conseguido el objetivo 1 (al menos 10000 pasos), el objetivo 2 (al menos 20000 pasos) y el objetivo 3 (al menos 30000 pasos). Observa que una persona puede cumplir varios objetivos; por ejemplo, si has dado 25000 pasos has cumplido los objetivos 1 y 2.

```python
pasos = int(input("Introduce el número de pasos que has dado hoy: "))

if pasos >= 10000:
    print("- ¡Conseguido el Objetivo 1! (Al menos 10,000 pasos)")

if pasos >= 20000:
    print("- ¡Conseguido el Objetivo 2! (Al menos 20,000 pasos)")

if pasos >= 30000:
    print("- ¡Conseguido el Objetivo 3! (Al menos 30,000 pasos)")
```

El programa anterior se puede hacer más eficiente si colocamos sentencias `if` dentro de sentencias `if` (si no has superado los 10000 pasos, no puedes haber superado los 20000 pasos). A esto se le llama control **anidado**:

```python
pasos = int(input("Introduce el número de pasos que has dado hoy: "))

if pasos >= 10000:
    print("- ¡Conseguido el Objetivo 1! (Al menos 10,000 pasos)")
    if pasos >= 20000:
        print("- ¡Conseguido el Objetivo 2! (Al menos 20,000 pasos)")
        if pasos >= 30000:
            print("- ¡Conseguido el Objetivo 3! (Al menos 30,000 pasos)")
```

En este caso tenemos un anidamiento de nivel tres (un `if`, dentro de un `if`, dentro de un `if`). Aunque Python no limita el nivel de anidamiento, por legibilidad se aconseja no superar el nivel tres (algunas normas de estilo incluso lo reducen a dos).

## El método `split()`

Hasta ahora solo hemos utilizado funciones como `print()`,  `input()`, `len()`:

```python
cadena = "hola"
print(len(cadena))
```

En Python, además de funciones, también se pueden usar **métodos**. Los métodos son _similares_ a las funciones, pero se usan de forma distinta.

Por ejemplo el método `upper` se aplica a una cadena y devuelve la cadena convertida a mayúsculas:

```python
"hola".upper()
```

```python
cadena.upper()
```

```python
cadena.len()
```

Observa que la sintaxis es `cadena.upper()`; no `upper(cadena)`.

Un método es similar a la función, pero el valor al que se aplica no aparece entre paréntesis; aparece antes del nombre del método, separado de este por un punto:

|            | Sintaxis           | Ejemplo           |
|------------|--------------------|-------------------|
|**Función** | nombre(expresion)  | `len("python")`   |
|**Método**  | expresion.nombre() | `"python".upper()`|

**Nota:** Lo que llamamos método es un realmente un _mensaje_. Los veremos al estudiar clases y objetos.

Observa que si `upper()` se aplica a una variable de tipo `str`, el valor de la variable no se modifica:

```python
print(cadena)
print(cadena.upper())
print(cadena)
```

El método `split()` separa una cadena en varias cadenas, usando _blancos_ como separadores:

```python
"Estamos en clase de programación con Python".split()
```

```python
"Hola:\tque tal\tadios".split()
```

También se puede indicar entre paréntesis qué carácter queremos usar como separador:

```python
"19/junio/2026".split("/")
```

El método `split()` se suele usar con la función `input()` para leer varios datos por teclado con un solo `input()`:

```python
x = input("x:")
y = input("y:")
z = input("z:")

print(x,y,z)
```

```python
x,y,z = input("x,y,z: ").split(",")
print(x,y,z)
```

```python
"19/junio/2026".split("/")
dd,mm,aa = input("dia,mes,año")
```
