# Unidad 16 - Conjuntos

Los conjuntos son colecciones mutables sin elementos repetidos. Al igual que los dicionarios (que son conjuntos de pares) los conjuntos no son secuencias. Esto significa que los elementos no ocupan una posición concreta y, por tanto, no se pueden indexar.

Un literal de conjunto tiene la siguiente sintaxis:

      {exp_0, exp_1, exp_2, ..., exp_n}

Es decir, aparecen los valores que contiene el conjunto separados por comas y encerrados entre corchetes:

```python
primos = { 2, 3, 5, 7, 11, 13, 17 }
primos
```

```python
colores = { "rojo", "verde", "azul", "amarillo" }
colores
```

El tipo de los elementos de un conjunto está limitado: los elementos deben ser de tipos inmutables (`int` ,`float`, `str`,...):

```python
valido = { True, "rojo", 2 * 3, 15 / 6 }  ## válido
valido
```

```python
no_valido = { True, "rojo", 2 * 3, 15 / 6, [1, 2, 3] }  ## no válido: no puede contener una lista
```

El tipo de los conjuntos se llama `set`, y no refleja ni su tamaño ni el tipo de sus elementos:

```python
print(type(primos))
print(type(colores))
print(type(valido))
```

Para construir un conjunto vacío debemos utilizar la función `set()` (el literal `{}` rerpesenta un diccionario vacío):

```python
vacio = set()
print(type(vacio))
print(type({}))
vacio
```

## Operaciones sobre conjuntos

Los conjuntos soportan la función `len()` y los operadores `in` y `not in`:

```python
print(primos)
print(len(primos))
print(2 in primos)
print(4 in primos)
```

## Unión, intersección y diferencia de conjuntos

Los conjuntos soportan tres operadores muy importantes:

- **unión**: `A | B` es el conjunto formado por los elementos que están en `A` o en `B`
- **intersección**: `A & B` es el conjunto formado por los elementos que están en `A` y en `B` (i.e. los comunes)
- **diferencia**: `A - B` es el conjunto formado por los elementos de `A` que no están en `B`

```python
a = { 1, 2, 3, 4, 5, 7, 8, 9, 10 }
b = { 0, 2, 4, 6, 8, 10}

print("a =", a)
print("b =", b)
print()
print("a | b =", a | b)
print("a & b =", a & b)
print("a - b =", a - b)
print("b - a =", b - a)
```

## Métodos sobre conjuntos

Los conjuntos soportan, entre otros, los siguientes métodos:
    
| Método         | Significado                       |
|----------------|-----------------------------------|
| `s.add(x)`     | añade `x` a `s`                   |
| `s.remove(x)`  | elimina `x` de `s`                |
| `s.discard(x)` | elimina `x` de `s`                |
| `s.clear()`    | elimina todos los elementos de `s`|

```python
s = set(range(0,10))
print(s)
s.add(-1000)
s.add(0)
print(s)
```

```python
print(s)
s.remove(5)  # ¿qué pasa si evaluamos dos veces esta celda?
print(s)
```

```python
print(s)
s.discard(2)  # ¿qué pasa si evaluamos dos veces esta celda?
print(s)
```

```python
print(s)
s.clear()
print(s)
```

## Iteración sobre conjuntos

Podemos utilizar un conjunto como fuente de datos en un bucle `for`:

```python
primos = { 2, 3, 5, 7, 11, 13, 17 }

for p in primos:
    print(p)
```

**Ejercicio 1**: Crear un conjunto con todos los números menores que 100 que sean múltiplos de 3. Mostrar los elementos por pantalla

```python
multiplos_de_3 = []

for numero in range(3,100):
    if numero % 3 == 0:
        multiplos_de_3.append(numero)

print("Múltiplos de 3 menores que 100:", multiplos_de_3)
```

**Ejercicio 2**: Crear un conjunto vacío y añadir los elementos pertenecientes a la lista [5,7,5,1,-3]. Mostrar los elementos del conjunto de mayor a menor.



```python
conjunto = []
lista_original = [5, 7, 5, 1, -3]

for elemento in lista_original:
    if elemento not in conjunto:
        conjunto.append(elemento)

conjunto.sort(reverse=True)

print(f"Elementos de mayor a menor:", conjunto)
```

**Ejercicio 3**: Dada una lista de elementos que pueden estar repetidos, utilizar un conjunto para calcular cuántos elementos distintos hay.

Ejemplos:

La lista [1, 4, -2, 3, 3, 3, -2] tiene 4 elementos únicos

La lista [1, 1, 1, 1, 1] tiene 1 elemento único

```python
lista_elementos = [1, 4, -2, 3, 3, 3, -2]

elementos_unicos = []

for elemento in lista_elementos:
    if elemento not in elementos_unicos:
        elementos_unicos.append(elemento)

cantidad_unicos = len(elementos_unicos)

print(f"La lista original tiene {cantidad_unicos} elementos únicos")
```

**Ejercicio 4**: Dada una lista de números, utiliza conjuntos para mostrar únicamente los elementos que aparecen más de una vez.

```python
lista_numeros = [1, 5, 3, 5, 2, 8, 1, 9, 3] 

vistos = []
repetidos = []

for numero in lista_numeros:
    if numero in vistos and numero not in repetidos:
        repetidos.append(numero)
    
    elif numero not in vistos:
        vistos.append(numero)

print("Elementos que aparecen más de una vez:", repetidos)
```

**Ejercicio 5**: En un instituto se imparten tres asignaturas de programación:

python = {"Ana", "Luis", "Pedro", "María", "Elena"} \
java = {"Pedro", "María", "Carlos", "Luis"} \
javascript = {"María", "Elena", "Sergio"}

Realizar las siguientes operaciones:
1. Obtener todos los alumnos matriculados en al menos una asignatura.
2. Obtener los alumnos matriculados tanto en Python como en Java.
3. Obtener los alumnos matriculados en Python pero no en JavaScript.
4. Obtener los alumnos que están matriculados en una única asignatura.

```python
python = {"Ana", "Luis", "Pedro", "María", "Elena"}
java = {"Pedro", "María", "Carlos", "Luis"}
javascript = {"María", "Elena", "Sergio"}

al_menos_una = python | java | javascript
print("1. Alumnos en al menos una asignatura:", al_menos_una)

python_y_java = python & java
print("2. Alumnos en Python y Java:", python_y_java)

python_no_js = python - javascript
print("3. Alumnos en Python pero no en JavaScript:", python_no_js)

solo_python = python - java - javascript
solo_java = java - python - javascript
solo_js = javascript - python - java

unica_asignatura = solo_python | solo_java | solo_js
print("4. Alumnos en una única asignatura:", unica_asignatura)
```
