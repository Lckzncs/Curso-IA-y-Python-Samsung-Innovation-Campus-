## 1. Multiplicación rusa

Escribe un programa que multiplique dos números naturales mediante el algoritmo de la multiplicación rusa. Este algoritmo multiplica repetitivamente por dos el multiplicando y divide por dos el multiplicador, hasta que el multiplicador toma el valor 0. La suma todos los multiplicandos correspondientes a multiplicadores impares es el producto de los números originales.
Por ejemplo: $37 \times 12 = 12 + 48 + 384 = 444$, como se observa en la siguiente tabla:

| Multiplicador | Multiplicando |
| ------------- | ------------- |
|      37       |    **12**     |
|      18       |      24       |
|       9       |    **48**     |
|       4       |      96       |
|       2       |     192       |
|       1       |   **384**     |



````python
multiplicador = int(input("Introduce el multiplicador:"))
multiplicando = int(input("Introduce el multiplicando:"))

resultado = 0

while multiplicador >= 1:
    if multiplicador % 2 != 0:
        resultado += multiplicando
        
    multiplicador = multiplicador // 2
    multiplicando = multiplicando * 2
    print(multiplicador, multiplicando)

print(f"El resultado es: {resultado}")
````

## 2. Suma de las cifras de un número

Escribe un programa que sume las cifras de un número. Por ejemplo, para $5127$ debe devolver $15$, ya que $5 + 1 + 2  +7 = 15$.

Sugerencia: dado un número, puedes utilizar los operadores `%` y `//` para obtener  su última cifra y todas sus cifras menos la última.

````python
numero = int(input("Introduce un numero:"))

numero_original = numero
suma = 0

while numero > 0:
    ultima_cifra = numero % 10
    suma += ultima_cifra
    resto_cifras = numero // 10
    print(f"La ultima cifra de {numero_original} es {ultima_cifra}, es resto de cifras es {resto_cifras}")
    numero = resto_cifras
print(f"La suma de las cifras de {numero_original} es: {suma}")
````

## 3. Fizz Buzz

Fizz Buzz es un juego para enseñar a los niños a dividir mentalmente. En clase los niños van contando por turnos desde 1 (o desde otro número inicial). Si el número que le toca en su turno es divisible por 3, el niño debe decir **Fizz**; si es divisible por 5, debe decir **Buzz**; si es divisible por 3 y por 5, debe decir **Fizz Buzz**; si no es divisible ni por 3 ni por 5, debe decir el número que le toque. Por ejemplo, la secuencia para los números desde el 1 hasta el 36 es la siguiente: 

```
1, 2, Fizz, 4, Buzz, Fizz, 7, 8, Fizz, Buzz, 11, Fizz, 13, 14, Fizz Buzz, 16, 17, Fizz, 19, Buzz, Fizz, 22, 23, Fizz, Buzz, 26, Fizz, 28, 29, Fizz Buzz, 31, 32, Fizz, 34, Buzz, Fizz, ...
```

Escribe un programa  que ayude a los niños a contar desde un número inicial dado hasta que hayan dicho un número de veces **Fizz**, **Buzz** o **FIzz Buzz**.

````python
numero_inicial = int(input("Introduce un numero:"))
limite_palabras = int(input("Introduce el numero de veces que aparece Fizz, Buzz ó Fizz Buzz:"))
contador_palabras = 0
numero_actual = numero_inicial

while contador_palabras < limite_palabras:
    if numero_actual % 3 == 0 and numero_actual % 5 == 0:
        print("Fizz Buzz")
        contador_palabras += 1
    elif numero_actual % 3 == 0:
        print("Fizz")
        contador_palabras += 1
    elif numero_actual % 5 == 0:
        print("Buzz")
        contador_palabras += 1
    else:
        print(numero_actual)
        
    numero_actual += 1
````

## 4. Cambio de base a binario

Escribe un programa que dado un entero en notación decimal escriba en pantalla su representación en binario.

````python
numero_decimal = int(input("Introduce un numero decimal:"))
numero_actual = numero_decimal
binario = ""

if numero_actual == 0:
    binario = "0"
else:
    while numero_actual > 0:
        binario = str(numero_actual % 2) + binario
        numero_actual = numero_actual // 2

print(f"El número {numero_decimal} en binario es: {binario}")
````

## 5. Cambio de base a decimal

Escribe un programa que dado un entero en binario lo pase a decimal, escribiendo su representación en pantalla.

````python
numero_binario = int(input("Introduce un numero binario:"))
numero_actual = numero_binario
decimal = 0
potencia = 0

while numero_actual > 0:
    ultimo_digito = numero_actual % 10
    decimal += ultimo_digito * (2 ** potencia)
    numero_actual = numero_actual // 10
    potencia += 1

print(f"El número binario {numero_binario} en decimal es: {decimal}")
````
