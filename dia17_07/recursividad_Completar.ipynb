# Recursividad en Python

## ¿Qué es la recursividad?

La **recursividad** es una técnica de programación en la que una función **se llama a sí misma** para resolver un problema. Es útil cuando un problema puede dividirse en subproblemas más pequeños y similares al problema original.

Para que una función recursiva funcione correctamente, debe tener al menos dos partes:

1. **Caso base**: la condición que **detiene** las llamadas recursivas.
2. **Paso recursivo**: la parte en la que la función **se llama a sí misma** con un valor más pequeño (más cerca del caso base).

## Ejemplo: Cálculo del factorial

El **factorial** de un número entero positivo $n$ (escrito como $n!$) se define así:

$$
n! = n \times (n - 1) \times (n - 2) \times \dots \times 1
$$

Por ejemplo:

$$
4! = 4 \times 3 \times 2 \times 1 = 24
$$

También se define que:

$$
0! = 1
$$

```python
# Versión recursiva del factorial
def factorial_recursivo(n):
    if n == 0:
        return 1  # Caso base
    else:
        return n * factorial_recursivo(n - 1)  # Paso recursivo
```

```python
# Ejemplo de uso
print(f"Factorial recursivo de 4: {factorial_recursivo(4)}")  # Imprime 24
```

```python
# Versión iterativa del factorial
def factorial_iterativo(n):
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    return resultado
```

```python
# Ejemplo de uso
print(f"Factorial iterativo de 4: {factorial_iterativo(4)}")  # Imprime 24
```

## Comparación entre versiones

| Característica     | Versión recursiva                 | Versión iterativa                  |
|--------------------|-----------------------------------|------------------------------------|
| Uso de funciones   | Se llama a sí misma               | Usa un bucle (`for` o `while`)     |
| Memoria            | Puede usar más memoria (pila)     | Más eficiente en memoria           |
| Claridad           | Puede ser más clara y elegante    | A veces más fácil de seguir        |
| Riesgo             | Puede causar error si hay muchas llamadas recursivas (stack overflow) | No tiene ese riesgo |

## ¿Cuándo usar recursividad?

- Cuando el problema se puede definir naturalmente de forma recursiva (como factorial, Fibonacci, recorrer árboles, etc.).
- Cuando mejora la legibilidad del código, aunque a veces no sea lo más eficiente.

## Ejercicios

Define funciones recursivas que resuelvan los siguientes problemas:

**1.** Sumar los primeros n números naturales: 1 + 2 + 3 + ... + n.

```python
def sumar_naturales(n):
    if n <= 1:
        return n
    return n + sumar_naturales(n - 1)

print(sumar_naturales(5))
```

**2.** Calcular la potencia de un número, base^exponente, sin usar **

```python
def potencia(base, exponente):
    if exponente == 0:
        return 1
    return base * potencia(base, exponente - 1)

print(potencia(2, 3))
```

**3.** Contar cuántos dígitos tiene un número

```python
def contar_digitos(n):
    n = abs(n)
    if n < 10:
        return 1
    return 1 + contar_digitos(n // 10)

print(contar_digitos(12345))
```

**4.** Invertir una cadena

```python
def invertir_cadena(cadena):
    if len(cadena) == 0:
        return cadena
    return invertir_cadena(cadena[1:]) + cadena[0]

print(invertir_cadena("hola"))
```

**5.** Devolver el n-ésimo término de la sucesión de Fibonacci: 0,1,1,2,3,5,8,...

```python
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(40))
```

**6.** Calcule el máximo común divisor mediante el algoritmo de Euclides.

 Sean $a$ y $b$ dos números enteros con $a ≥ b > 0$
 
 Dividir $a$ entre $b$ y obtener el residuo $r$ tal que $a=b·q+r$.

 Sustituir $a$ por $b$ y $b$ por $r$

 Repetir el proceso hasta que el residuo sea $0$.

 El último valor de $b$ distinto de cero es el MCD

 **Ejemplo:**

270 ÷ 192 = 1 residuo 78 → MCD(270,192) = MCD(192,78)

192 ÷ 78 = 2 residuo 36 → MCD(192,78) = MCD(78,36)

78 ÷ 36 = 2 residuo 6 → MCD(78,36) = MCD(36,6)

36 ÷ 6 = 6 residuo 0 → MCD = 6 


```python
def mcd(a, b):
    if b == 0:
        return a
    return mcd(b, a % b)

print(mcd(270, 192))
```

**7.** Verificar si una cadena es un palíndromo. En caso de existir espacios deben ignorarse.

```python
def es_palindromo(cadena):
    cadena = cadena.replace(" ", "").lower()
    
    if len(cadena) <= 1:
        return True
        
    if cadena[0] == cadena[-1]:
        return es_palindromo(cadena[1:-1])
        
    return False

print(es_palindromo("anita lava la tina"))
```

**8.** Sumar los dígitos de un número

```python
def sumar_digitos(n):
    n = abs(n)
    if n < 10:
        return n
    return (n % 10) + sumar_digitos(n // 10)

print(sumar_digitos(1234))
```

**9.** Búsqueda lineal recursiva.  Devuelve la posición del elemento *objetivo* en la sublista $lista[inicio:fin+1]$. Si no se encuentra se debe devolver -1. 

```python
def busqueda_lineal(lista, objetivo, indice=0):
    if indice == len(lista):
        return -1
    
    if lista[indice] == objetivo:
        return indice
        
    return busqueda_lineal(lista, objetivo, indice + 1)

print(busqueda_lineal([10, 5, 3, 8, 2], 8))
```

**10.** Búsqueda binaria recursiva. Devuelve la posición del elemento *objetivo* en la sublista $lista[inicio:fin+1]$. Si no se encuentra se debe devolver -1. 

```python
def busqueda_binaria(lista, objetivo, inicio=0, fin=None):
    if fin is None:
        fin = len(lista) - 1

    if inicio > fin:
        return -1

    medio = (inicio + fin) // 2

    if lista[medio] == objetivo:
        return medio
        
    if lista[medio] > objetivo:
        return busqueda_binaria(lista, objetivo, inicio, medio - 1)
        
    return busqueda_binaria(lista, objetivo, medio + 1, fin)

print(busqueda_binaria([2, 4, 6, 8, 10, 12, 14], 10))
```
