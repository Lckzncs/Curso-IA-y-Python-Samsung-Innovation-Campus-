## Ecuaciones en markdown (LaTeX)

Para escribir ecuaciones usamos $\LaTeX$.

El área $a$ de un círculo de radio $r$ es $a = \pi  r^2$.

$$a = \pi  r^2$$

$$a = \pi \cdot r^2$$

$$a = \pi \times r^2$$

Para las fracciones usamos `\frac`:

$$
\frac{x^2 - 3 x}{\sin x}
$$

````python
# Esta es una celda de código.
# Para ejecutarla presionamos Ctr- Intro

from matplotlib import pyplot as plt
import numpy as np

# Generate 80 random data points along 3 dimensions
x, y, scale = np.random.randn(3, 80)
fig, ax = plt.subplots()

# Map each onto a scatterplot we'll create with Matplotlib
ax.scatter(x=x, y=y, c=scale, s=np.abs(scale)*200)
ax.set(title="Some random data, created with JupyterLab!")
plt.show()
````
