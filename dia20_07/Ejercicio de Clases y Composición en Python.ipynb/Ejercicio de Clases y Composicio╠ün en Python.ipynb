## Ejercicio de Clases y Composición en Python

Crea un programa en Python que gestione una empresa y sus empleados. Para ello, deberás implementar las siguientes clases:

### Clase Empleado
Esta clase representa a un empleado de la empresa y tiene los siguientes atributos privados:

`dni` (string): el número de identificación del empleado.

`nombre` (string): el nombre del empleado.

`edad` (int): la edad del empleado.

La clase Empleado debe tener los siguientes métodos:

`__init__`(self, dni, nombre, edad): constructor que recibe los atributos dni, nombre y edad y los asigna a los respectivos atributos privados.

`__eq__`(self, other): método que compara si dos empleados son iguales en base a su número de DNI.

`__str__`(self): método que devuelve una representación en cadena del empleado.

Además, se deberán definir los getters y setters encapsulados, preferiblemente con decoradores.


### Clase Empresa
Esta clase representa una empresa y tiene los siguientes atributos privados:

`nombre` (string): el nombre de la empresa.

`empleados` (list): una lista que almacenará objetos de la clase Empleado.


La clase Empresa debe tener los siguientes métodos:

`__init__`(self, nombre): constructor que recibe el atributo nombre y lo asigna al correspondiente atributo privado.

`agregar_empleado`(self, empleado): método que recibe un objeto de la clase Empleado y lo agrega a la lista de empleados si el empleado no está. En caso de que esté, se muestra un mensaje por pantalla.

`borrar_empleado`(self, dni): método que elimina un empleado de la lista según su número de DNI. Si el empleado no está, se muestra un mensaje por pantalla. 

`__str__`(self): método que devuelve una representación en cadena de la empresa y sus empleados.

`guardar_empleados`(self, archivo): método que guarda la lista de empleados en un archivo utilizando la librería pickle.

`cargar_empleados`(self, archivo): método que carga la lista de empleados desde un archivo utilizando la librería pickle.

Para probar el programa, crea objetos de la clase Empleado, agrega empleados a una instancia de la clase Empresa, realiza operaciones como agregar y borrar empleados, y guarda y carga los empleados en un archivo.

```python
# Definición de las clases

class Empleado:
    def __init__(self, dni, nombre, edad):
        self.__dni = dni
        self.__nombre = nombre
        self.__edad = int(edad)

    @property
    def dni(self):
        return self.__dni

    @dni.setter
    def dni(self, nuevo_dni):
        self.__dni = nuevo_dni

    @property
    def nombre(self):
        return self.__nombre

    @nombre.setter
    def nombre(self, nuevo_nombre):
        self.__nombre = nuevo_nombre

    @property
    def edad(self):
        return self.__edad

    @edad.setter
    def edad(self, nueva_edad):
        self.__edad = int(nueva_edad)

    def __eq__(self, otro):
        if isinstance(otro, Empleado):
            return self.__dni.lower == otro.dni.lower
        return False

    def __str__(self):
        return f"[DNI: {self.__dni}] {self.__nombre}, {self.__edad} años"


class Empresa:
    def __init__(self, nombre):
        self.__nombre = nombre
        self.__empleados = []

    def agregar_empleado(self, empleado):
        if empleado in self.__empleados:
            print(f"El empleado con DNI {empleado.dni} ya trabaja aquí.")
        else:
            self.__empleados.append(empleado)

    def borrar_empleado(self, dni):
        empleado_a_borrar = None
        
        for emp in self.__empleados:
            if emp.dni == dni and empleado_a_borrar is None:
                empleado_a_borrar = emp
        
        if empleado_a_borrar is not None:
            self.__empleados.remove(empleado_a_borrar)
            print(f"Empleado con DNI {dni} borrado correctamente.")
        else:
            print(f"No hemos encontrado a nadie con el DNI {dni}.")

    def __str__(self):
        resultado = f"--- Empresa: {self.__nombre} ---\n"
        if not self.__empleados:
            resultado += "Todavía no hay empleados registrados."
        else:
            for emp in self.__empleados:
                resultado += f"- {emp}\n"
        return resultado

    def guardar_empleados(self, archivo):
        with open(archivo, 'w', encoding='utf-8') as f:
            for emp in self.__empleados:
                f.write(f"{emp.dni},{emp.nombre},{emp.edad}\n")

    def cargar_empleados(self, archivo):
        with open(archivo, 'r', encoding='utf-8') as f:
            self.__empleados = []
            for linea in f:
                datos = linea.strip().split(',')
                if len(datos) == 3:
                    nuevo_empleado = Empleado(datos[0], datos[1], datos[2])
                    self.__empleados.append(nuevo_empleado)
```

```python
# Uso de las clases

mi_empresa = Empresa("Tech Solutions")

emp1 = Empleado("12345678A", "Ana García", 30)
emp2 = Empleado("87654321B", "Carlos López", 45)
emp3 = Empleado("11223344C", "Beatriz Ruiz", 28)

print("Añadiendo a las primeras personas...")
mi_empresa.agregar_empleado(emp1)
mi_empresa.agregar_empleado(emp2)
mi_empresa.agregar_empleado(emp3)

print("\nProbando a meter a alguien repetido:")
mi_empresa.agregar_empleado(Empleado("12345678A", "Ana García (Clon)", 30))

print("\nAsí está la empresa ahora mismo:")
print(mi_empresa)

print("Guardando los datos en 'datos.txt'...")
mi_empresa.guardar_empleados('datos.txt')

print("\nDespidiendo a Carlos...")
mi_empresa.borrar_empleado("87654321B")

print("\nLa empresa tras la salida de Carlos:")
print(mi_empresa)

print("Recuperando la lista desde 'datos.txt'...")
mi_empresa.cargar_empleados('datos.txt')

print("\nLa empresa después de cargar los datos antiguos:")
print(mi_empresa)
```
