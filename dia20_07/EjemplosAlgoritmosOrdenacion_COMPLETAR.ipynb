**1.** Modificar las versiones de los algoritmos de ordenación que actualizan la lista proporcionada como entrada para que devuelvan el número de intercambios que hace cada algoritmo. 

```python
def burbuja_intercambios(lista):
    n = len(lista)
    intercambios = 0
    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
                intercambios += 1
    return intercambios

def selectionsort_intercambios(S):
    n = len(S)
    intercambios = 0
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if S[min_idx] > S[j]:
                min_idx = j
        
        if i != min_idx:
            S[i], S[min_idx] = S[min_idx], S[i]
            intercambios += 1
    return intercambios

def insertionsort_intercambios(S):
    intercambios = 0
    for i in range(1, len(S)):
        j = i
        while j > 0 and S[j - 1] > S[j]:
            S[j - 1], S[j] = S[j], S[j - 1]
            intercambios += 1
            j -= 1
    return intercambios

S1 = [50, 30, 40, 10, 20]
print(S1)
print(f"Intercambios Burbuja S1: {burbuja_intercambios(S1)}")

S1 = [50, 30, 40, 10, 20]
print(f"Intercambios Selección S1: {selectionsort_intercambios(S1)}")

S1 = [50, 30, 40, 10, 20]
print(f"Intercambios Inserción S1: {insertionsort_intercambios(S1)}")
print(S1)

S2 = [10, 20, 30, 40, 50]
print(S2)
print(f"Intercambios Burbuja S2: {burbuja_intercambios(S2)}")

S2 = [10, 20, 30, 40, 50]
print(f"Intercambios Selección S2: {selectionsort_intercambios(S2)}")

S2 = [10, 20, 30, 40, 50]
print(f"Intercambios Inserción S2: {insertionsort_intercambios(S2)}")
print(S2)
```
