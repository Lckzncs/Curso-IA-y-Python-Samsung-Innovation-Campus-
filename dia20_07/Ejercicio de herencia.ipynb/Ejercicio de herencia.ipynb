## Ejercicio de Herencia

Crea un programa en Python que simule un sistema de astronautas. Deberás implementar las siguientes clases:

### Clase `Astronauta`
Esta clase representa a un astronauta genérico y tiene los siguientes atributos privados:

- `_nombre` (str): el nombre del astronauta.
- `_experiencia` (int): la experiencia en años del astronauta.

La clase `Astronauta` debe tener los siguientes métodos:

- `get_nombre(self)`: método que devuelve el nombre del astronauta.
- `set_nombre(self, nuevo_nombre)`: método que actualiza el nombre del astronauta.
- `get_experiencia(self)`: método que devuelve la experiencia del astronauta.
- `set_experiencia(self, nueva_experiencia)`: método que actualiza la experiencia del astronauta.
- `saludar(self)`: método que muestra un mensaje de saludo con el nombre y la experiencia del astronauta.
- `__str__(self)`: método que devuelve una representación en cadena del astronauta.
- `__eq__(self, other)`: método que compara si dos astronautas son iguales en base a su nombre y experiencia.

### Clase `AstronautaEspacial`
Esta clase hereda de `Astronauta` y representa a un astronauta especializado en misiones espaciales. Además de los atributos heredados, tiene un atributo privado adicional:

- `_misiones_realizadas` (int): el número de misiones espaciales realizadas por el astronauta.

La clase `AstronautaEspacial` debe tener los siguientes métodos:

- `get_misiones_realizadas(self)`: método que devuelve el número de misiones realizadas.
- `set_misiones_realizadas(self, nuevas_misiones)`: método que actualiza el número de misiones realizadas.
- `saludar(self)`: método que muestra un mensaje de saludo específico para astronautas espaciales.
- `__str__(self)`: método que devuelve una representación en cadena del astronauta espacial.
- `__eq__(self, other)`: método que compara si dos astronautas espaciales son iguales en base a su nombre, experiencia y número de misiones realizadas.

### Clase `AstronautaLunar`
Esta clase también hereda de `Astronauta` y representa a un astronauta especializado en caminatas lunares. Además de los atributos heredados, tiene un atributo privado adicional:

- `_caminatas_lunares` (int): el número de caminatas lunares realizadas por el astronauta.

La clase `AstronautaLunar` debe tener los siguientes métodos:

- `get_caminatas_lunares(self)`: método que devuelve el número de caminatas lunares realizadas.
- `set_caminatas_lunares(self, nuevas_caminatas)`: método que actualiza el número de caminatas lunares realizadas.
- `saludar(self)`: método que muestra un mensaje de saludo específico para astronautas lunares.
- `__str__(self)`: método que devuelve una representación en cadena del astronauta lunar.
- `__eq__(self, other)`: método que compara si dos astronautas lunares son iguales en base a su nombre, experiencia y número de caminatas lunares realizadas.

Para probar el programa, crea objetos de las clases `AstronautaEspacial` y `AstronautaLunar`, utiliza los métodos de encapsulación `get` y `set` para modificar los atributos, y muestra la información de los astronautas utilizando el método `__str__`. También realiza comparaciones entre los astronautas utilizando el método `__eq__`.
